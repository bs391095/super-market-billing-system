````python
#```python
# ============================================================
#                 GROCERY BILLING SYSTEM
# ============================================================

# ---------------- PRODUCT LIST ----------------

print("=" * 90)
print("                         AVAILABLE PRODUCTS")
print("=" * 90)

print(f"{'Product':<25} {'Product ID':<15} {'Category':<18} {'Price':>10}")
print("-" * 90)


# Dictionary containing all products
# Each product is stored as a tuple:
# (Product Name, Category, Price)

products = {
    1: ("Rice(1KG)", "Grocery", 60),
    2: ("Wheat flour(1KG)", "Grocery", 40),
    3: ("Milk(1L)", "Dairy", 60),
    4: ("Butter(100G)", "Dairy", 50),
    5: ("Cold Drink(500ml)", "Beverages", 40),
    6: ("Chips(40G)", "Snacks", 20),
    7: ("Biscuits(200G)", "Snacks", 30),
    8: ("Soap(125G)", "Personal_Care", 30),
    9: ("Shampoo(250ml)", "Personal_Care", 120),
    10: ("Detergent_Powder(1KG)", "HouseHold", 120)
}


# Display products
for product_id, product_details in products.items():

    # Tuple unpacking
    product_name, category, price = product_details

    print(f"{product_name:<25} "
          f"{'product_Id(' + str(product_id) + ')':<15} "
          f"{category:<18} "
          f"{price:>10}")


print("=" * 90)


# ============================================================
#                    SHOPPING / BILLING
# ============================================================

print("                              ITEMIZED BILL")
print("=" * 90)

print(f"{'Product':<25} {'Product ID':<15} {'Qty':<10} {'Amount':>12}")
print("-" * 90)


# Empty list to store purchased products
cart = []


# ---------------- PRODUCT SELECTION ----------------

while True:

    try:
        product_id = int(input("Enter the product ID: "))
        
        # Check whether product ID exists
        if product_id not in products:
            print("Invalid Product ID! Please enter a valid product ID.")
            continue

        qty = int(input("Enter the quantity: "))

        # Check quantity
        if qty <= 0:
            print("Quantity must be greater than 0.")
            continue

        # Get product information from dictionary
        product_name, category, price = products[product_id]

        # Calculate amount
        amount = price * qty

        # Create a tuple for purchased item
        item = (product_name, product_id, qty, price, amount)

        # Add item to the cart using append()
        cart.append(item)

        print(f"{product_name} x {qty} added successfully.")
        print(f"Amount: ₹{amount}")

        # Ask whether user wants another product
        user = input("Do you want to add more products? (y/n): ")

        if user.lower() == "n" or user.lower() == "no":
            break

    except ValueError:
        print("Please enter numbers only.")


# ============================================================
#                       ITEMIZED BILL
# ============================================================

print()
print("=" * 90)
print("                              FINAL BILL")
print("=" * 90)

print(f"{'Product':<25} {'Product ID':<15} {'Qty':<10} {'Amount':>12}")
print("-" * 90)


# Variable for subtotal
subtotal = 0


# Display all products from cart
for item in cart:

    # Tuple unpacking
    product_name, product_id, qty, price, amount = item

    print(f"{product_name:<25} "
          f"{product_id:<15} "
          f"{qty:<10} "
          f"₹{amount:>10}")

    # Add amount to subtotal
    subtotal = subtotal + amount


print("-" * 90)

print(f"{'SUBTOTAL':<65} ₹{subtotal:>10}")


# ============================================================
#                         DISCOUNT
# ============================================================

print()
print("=" * 90)
print("                         DISCOUNT OPTIONS")
print("=" * 90)

print("1. No Discount")
print("2. Flat Discount")
print("3. Percentage Discount")
print("-" * 90)


while True:

    try:
        discount_type = int(input("Enter your discount choice (1/2/3): "))

        if discount_type not in [1, 2, 3]:
            print("Please select 1, 2 or 3.")
            continue

        break

    except ValueError:
        print("Please enter 1, 2 or 3.")


# Default discount
discount = 0


# ---------------- NO DISCOUNT ----------------

if discount_type == 1:

    discount = 0

    print("No discount applied.")


# ---------------- FLAT DISCOUNT ----------------

elif discount_type == 2:

    while True:

        try:
            flat_discount = float(input("Enter flat discount amount: ₹"))

            if flat_discount < 0:
                print("Discount cannot be negative.")
                continue

            # Discount cannot be greater than subtotal
            if flat_discount > subtotal:
                print("Discount cannot be greater than the subtotal.")
                continue

            discount = flat_discount
            break

        except ValueError:
            print("Please enter a valid amount.")


# ---------------- PERCENTAGE DISCOUNT ----------------

elif discount_type == 3:

    while True:

        try:
            discount_percentage = float(
                input("Enter discount percentage: ")
            )

            if discount_percentage < 0 or discount_percentage > 100:
                print("Percentage must be between 0 and 100.")
                continue

            # Calculate percentage discount
            discount = subtotal * discount_percentage / 100

            break

        except ValueError:
            print("Please enter a valid percentage.")


# ============================================================
#                     FINAL CALCULATION
# ============================================================

final_amount = subtotal - discount


print()
print("=" * 90)
print("                         PAYMENT SUMMARY")
print("=" * 90)

print(f"{'Subtotal':<65} ₹{subtotal:>10}")
print(f"{'Discount':<65} ₹{discount:>10}")
print(f"{'Final Amount':<65} ₹{final_amount:>10}")

print("=" * 90)
print("                    THANK YOU FOR SHOPPING!")
print("=" * 90)
````
